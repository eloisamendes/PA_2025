Cálculo de média

calculomedia.drawio.png

```python
# Recebe as notas das provas
nota1 = float(input("Digite a nota da P1: "))
nota2 = float(input("Digite a nota da P2: "))

# Calcula a média
media = (nota1 + nota2) / 2

# Mostra a média
print(f"A média é: {media}")
```

Conversão moeda

conversaomoeda.drawio.png

```python
# Recebe a cotação do dólar e o valor em reais
cotacao_dolar = float(input("Digite a cotação do dólar: "))
valor_reais = float(input("Digite o valor em reais: "))

# Converte para dólares
valor_dolar = valor_reais / cotacao_dolar

# Mostra o resultado
print(f"O valor em dólares é: ${valor_dolar:.2f}")
```

Cálculo de comissão

comissao.drawio.png

```python
# Recebe o preço unitário e a quantidade vendida
preco_unitario = float(input("Digite o preço unitário da peça: "))
quantidade_vendida = int(input("Digite a quantidade vendida: "))

# Calcula o valor total da venda e a comissão
valor_total_venda = preco_unitario * quantidade_vendida
comissao = valor_total_venda * 0.05

# Mostra os resultados
print(f"Valor total da venda: R${valor_total_venda:.2f}")
print(f"Comissão do vendedor: R${comissao:.2f}")
```

Exercicio 1 - custo do carro

custocarro.drawio.png

```python
# Recebe o preço de fabrica
custo_fabrica = float(input("Digite o custo de fábrica do carro: R$ "))

# Calcula o valor final
impostos = custo_fabrica * 0.45
valor_com_impostos = custo_fabrica + impostos
distribuicao = valor_com_impostos * 0.28
custo_final = valor_com_impostos + distribuicao

# Recebe o resultado
print(f"\nCusto ao consumidor: R$ {custo_final:.2f}")
```

Exercicio 2 - Horas e minutos para segundos

conversaohoraeminuto.drawio.png

```python
# Solicitar ao usuário a hora e os minutos
hora = int(input("Digite a hora: "))
minutos = int(input("Digite os minutos: "))

# Calcular a fração dos minutos
minutos_decimais = minutos / 60

# Calcular a hora decimal
hora_decimal = hora + minutos_decimais

# Exibir o resultado
print(f"O valor em horas decimais é: {hora_decimal:.2f}")
```

Exercicio 3 - Hora extra

horaextra.drawio.png

```python
# Entrada
horas = int(input("Horas trabalhadas: "))
minutos = int(input("Minutos trabalhados: "))
valor_hora = float(input("Valor por hora: R$ "))

# Cálculo do salário com 70% de adicional
salario_total = (horas + minutos / 60) * valor_hora * 1.7

# Exibição do resultado
print(f"Total a receber: R$ {salario_total:.2f}")
```

Lista extra - Simulação de Investimento em CDB com Juros Compostos

listaextra.drawio.png

```python
# Entrada de dados
C = float(input("Digite o valor inicial do investimento: "))
i = float(input("Digite a taxa de juros anual (em %): ")) / 100
t = int(input("Digite o tempo de investimento (em anos): "))

# Cálculo do montante final usando a fórmula dos juros compostos
M = C * (1 + i) ** t

# Exibição do resultado
print(f"O montante final após {t} anos será de R$ {M:.2f}")

```
