```python
#Entrada
LN= [1, 2, 3, 4, 5, 6, 7]
QC= [8, 9, 10, 11, 12, 13, 14]
LC= [15, 16, 17, 18, 19, 20, 21]
QM= [22, 23, 24, 25, 26, 27, 28, 29]
NCL= [30, 31]

#Processamento

num = int(input("Digite o dia do mês: ")) 

if num in LN:
    print(f"O dia {num} terá lua nova!")
elif num in QC:
    print(f"O dia {num} terá quarto crescente!")
if num in LC:
    print(f"O dia {num} terá lua crescente!")
elif num in QM:
    print(f"O dia {num} terá quarto minguante!")
elif num in NCL:
    print(f"O dia {num} será o inicio de um novo ciclo lunar!")
```
