# **Impar ou par**

```python
#Entrada
numero = int(input("Digite um número: "))

#Processamento
paridade = ["Par", "Ímpar"][numero % 2]

#Saída
print(f"O número {numero} é {paridade}.")
```

# **Valores do triângulo**

```python
# Entrada
l1 = int(input("Digite o valor do primeiro lado: "))
l2 = int(input("Digite o valor do segundo lado: "))
l3 = int(input("Digite o valor do terceiro lado: "))

# Processamento
# Verificar se forma um triângulo
forma_triangulo = (l1 + l2 > l3) and (l1 + l3 > l2) and (l2 + l3 > l1)

# Se for um triângulo, identificar o tipo
equilatero = forma_triangulo and (l1 == l2 == l3)
isoceles = forma_triangulo and ((l1 == l2 and l1 != l3) or (l1 == l3 and l1 != l2) or (l2 == l3 and l2 != l1))
escaleno = forma_triangulo and (l1 != l2 and l1 != l3 and l2 != l3)

# Saída
print(f"Forma um triângulo? {forma_triangulo}")
print(f"É equilátero? {equilatero}")
print(f"É isósceles? {isoceles}")
print(f"É escaleno? {escaleno}")
```

# **Calculo de IMC**

```python
#Entrada
peso = float(input("Digite seu peso (kg): "))
altura = float(input("Digite sua altura (m): "))

#Processamento
imc = peso / (altura ** 2)

faixas = [
    (imc < 18.5, "Abaixo do peso"),
    (imc < 25, "Peso normal"),
    (imc < 30, "Sobrepeso"),
    (imc < 35, "Obesidade grau 1"),
    (imc < 40, "Obesidade grau 2"),
    (imc >= 40, "Obesidade grau 3")
]
status = next(texto for cond, texto in faixas if cond)

#Saida
print(f"Seu IMC é {imc:.2f}")
print(f"Classificação: {status}")
```
