# Exercícios

```python
# Entrada
nota1 = float(input('Digite a primeira nota: '))
nota2 = float(input('Digite a segunda nota: '))

# Processamento
media = (nota1 + nota2) / 2

# Saída
print('A média é :' , media)
```

Valor total:

```python
# Entrada
VALOR = float(input('Digite o valor: '))
QTDE = float(input('Digite a quantidade: '))

# Processamento
VT = (QTDE * VALOR)

# Saída
print ('O valor total é:', VT)
```

Triplo de um número:

```python
# Entrada
Número = float(input('Digite o número: '))

# Processamento
Triplo = (Número * 3)

# Saída
print ('O triplo do número é:' , Triplo)
```

Graus celsius para fahrenheit:

```python
# Entrada
Celsius = float(input('Digite a temperatura em graus celsius: '))

# Processamento
Fahrenheit = (32) + Celsius * 1.8

#Saída
print ('A temperatura em fahrenheit é' , Fahrenheit)
```

Milha para quilômetros:

```python
# Entrada
Milha = float(input('Digite a distância em milhas: '))

# Processamento
Quilômetros = (Milha * 1.609)

#Saída
print ('A distância em quilômetros é: ' , Quilômetros)
```
